```sql
-- Cellule 1 : Top clients par CA
SELECT
    c.nom_client,
    c.region,
    COUNT(f.order_id)           AS nb_commandes,
    SUM(f.montant_net)           AS chiffre_affaires,
    ROUND(AVG(f.remise) * 100, 1) AS remise_moy_pct
FROM       hive_metastore.gold.fact_ventes  f
INNER JOIN hive_metastore.gold.dim_client   c ON f.sk_client  = c.sk_client
GROUP BY   c.nom_client, c.region
ORDER BY   chiffre_affaires DESC;

-- Cellule 2 : CA par catégorie et mois
SELECT
    d.annee,
    d.mois,
    p.categorie,
    COUNT(*)                 AS nb_lignes,
    SUM(f.qte)               AS total_qte,
    SUM(f.montant_net)       AS ca_net
FROM       hive_metastore.gold.fact_ventes   f
INNER JOIN hive_metastore.gold.dim_date     d ON f.sk_date    = d.sk_date
INNER JOIN hive_metastore.gold.dim_produit  p ON f.sk_produit = p.sk_produit
GROUP BY   d.annee, d.mois, p.categorie
ORDER BY   d.annee, d.mois;
```
